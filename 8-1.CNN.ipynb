### 2. 채널(Channel)

* 이미지는 **(높이, 너비, 채널)** 이라는 3차원 텐서
    - 높이: 이미지의 세로 방향 픽셀 수
    - 너비: 이미지의 가로 방향 픽셀 수
    - 채널: 색 성분
* 흑백 이미지는 채널 수가 1, 컬러이미지는 적색(Red), 녹색(Green), 청색(Blue)으로 채널 수가 3개

### 3. 합성곱 연산(Convolution operation)

* 합성곱층은 합성곱 연산을 통해서 **이미지의 특징을 추출**
* 합성곱: **커널(kernel)** 또는 **필터(filter)** 라는 $n \times m$ 크기의 행렬로 높이 $\times$ 너비 크기의 이미지를 처음부터 끝까지 겹치며 훑으면서 $n \times m$ 크기의 겹쳐지는 부분의 각 이미지와 커널의 원소의 값을 곱해서 모두 더한 값을 출력으로 하는 것
    - 커널(kernel)은 일반적으로 $3 \times 3$ 또는 $5 \times 5$를 사용

### 5. 출력 크기 계산

합성곱 연산 후 특성 맵의 크기는 다음과 같이 계산

$$
O = \frac{I - K + 2P}{S} + 1
$$

* $I$: 입력 크기, $K$: 커널 크기, $P$: 패딩, $S$: 스트라이드
* 위 예시 (입력 5x5, 커널 3x3)
    - 스트라이드 1, 패딩 0: $(5 - 3 + 0) / 1 + 1 = 3$ → 3x3
    - 스트라이드 2, 패딩 0: $(5 - 3 + 0) / 2 + 1 = 2$ → 2x2
    - 스트라이드 1, 패딩 1: $(5 - 3 + 2) / 1 + 1 = 5$ → 5x5 (입력과 같은 크기 유지)

### 8. 풀링(Pooling)

* 합성곱 층 (합성곱 연산 + 활성화 함수) 다음에 풀링 층을 추가하는 것이 일반적
* 풀링(Pooling): 특성 맵을 다운샘플링하는 연산
    - 최대 풀링(max pooling)과 평균 풀링(average pooling) 등을 사용
* 공간 차원 감소, 이미지의 중요 정보만 추출

```python
# 스트라이드 2, 패딩 0, MAX POOLING
Image(url='https://wikidocs.net/images/page/62306/maxpooling.PNG', width=400)
```

* 풀링도 출력 크기 공식이 같음 (풀링 크기를 $K$로 보고 계산)
    - 입력 28x28, 풀링 크기 2, 스트라이드 2: $(28 - 2) / 2 + 1 = 14$ → 14x14
    - 크기가 절반으로 줄어듦
* 풀링에는 학습할 가중치가 없음

### 9. 파이토치로 확인하기

앞에서 배운 합성곱과 풀링을 파이토치 코드로 확인

#### 1) 입력 모양

* 파이토치의 합성곱은 입력을 **(배치 크기, 채널, 높이, 너비)** 형태의 4차원 텐서로 받음
    - 배치 크기: 한 번에 넣는 이미지의 수
    - 채널: 흑백 이미지는 1, 컬러 이미지는 3
    - 예) 5x5 흑백 이미지 1장 → `(1, 1, 5, 5)`
* 2장에서 본 (높이, 너비, 채널)과 달리 파이토치는 **채널을 높이, 너비보다 앞에** 둠
* 커널도 **(출력 채널, 입력 채널, 높이, 너비)** 형태의 4차원 텐서
    - 7장에서 본 것처럼 커널의 채널 수는 입력의 채널 수와 같아야 함
    - 출력 채널 수 = 커널의 개수 (커널 하나가 특성 맵 하나를 만듦)

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

#### 2) 합성곱 연산 확인

* 3장에서 손으로 계산한 예시를 `F.conv2d(입력, 커널)`로 계산해서 같은 특성 맵이 나오는지 확인
* 5x5 흑백 이미지 1장 → `(1, 1, 5, 5)`, 3x3 커널 1개 → `(1, 1, 3, 3)`

```python
# 3장 예시의 입력 (5x5)
x = torch.FloatTensor([[1, 2, 3, 4, 5],
                       [2, 1, 0, 1, 2],
                       [3, 0, 1, 1, 0],
                       [1, 4, 1, 1, 2],
                       [2, 1, 1, 0, 0]])

# 3장 예시의 커널 (3x3)
kernel = torch.FloatTensor([[1, 0, 1],
                            [1, 0, 1],
                            [0, 1, 0]])

# 4차원 텐서로 변환
x = x.view(1, 1, 5, 5)
kernel = kernel.view(1, 1, 3, 3)
```

```python
out = F.conv2d(x, kernel)
print(out)
print(out.shape)
```

#### 3) 스트라이드와 패딩

5장의 공식으로 계산한 출력 크기와 같은지 확인

```python
# 스트라이드 2
out = F.conv2d(x, kernel, stride=2)
print(out)
print(out.shape)
```

```python
# 패딩 1
out = F.conv2d(x, kernel, padding=1)
print(out)
print(out.shape)
```

#### 4) `nn.Conv2d`로 합성곱 층 만들기

* 위에서는 커널 값을 직접 정했지만, 실제 신경망에서는 커널 값이 **학습을 통해 정해짐**
* `nn.Conv2d(in_channels, out_channels, kernel_size, stride, padding)`로 합성곱 층을 만들면 커널 값은 랜덤으로 초기화됨
    - `in_channels`: 입력 채널 수, `out_channels`: 출력 채널 수 (= 커널의 개수)
* 커널 값은 랜덤이므로 출력 값 대신 **출력 크기**만 확인

```python
# MNIST 이미지 1장과 같은 크기의 입력 (배치 크기 1, 채널 1, 28x28)
image = torch.randn(1, 1, 28, 28)

conv = nn.Conv2d(in_channels=1, out_channels=32, kernel_size=3, stride=1, padding=1)
out = conv(image)
print(out.shape)  # (28 - 3 + 2) / 1 + 1 = 28
```

##### 연습

아래 설정으로 합성곱을 했을 때 출력 크기를 먼저 손으로 계산해 보고, 코드로 확인해 보기

1. 입력 28x28, 커널 3x3, 스트라이드 1, 패딩 0
2. 입력 28x28, 커널 5x5, 스트라이드 1, 패딩 2
3. 입력 28x28, 커널 4x4, 스트라이드 2, 패딩 1

```python
# 1번
conv = nn.Conv2d(in_channels=1, out_channels=32, kernel_size=3, stride=1, padding=0)
print(conv(image).shape)

# 2번, 3번도 직접 작성해 보기
```

#### 5) 풀링 확인

8장의 최대 풀링(풀링 크기 2, 스트라이드 2)을 합성곱 뒤에 적용하면 크기가 절반으로 줄어듦

```python
conv = nn.Conv2d(in_channels=1, out_channels=32, kernel_size=3, stride=1, padding=1)
pool = nn.MaxPool2d(kernel_size=2, stride=2)

out = conv(image)
print('합성곱 후:', out.shape)

out = pool(out)
print('풀링 후:', out.shape)
```
