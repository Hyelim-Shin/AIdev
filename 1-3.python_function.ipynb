### 2) 여러 개의 매개변수

```python
def 함수_이름(*매개변수):
    수행할_문장
```

````python
def add_many(*args):
    result = 0
    for i in args:
        result += i
    return result
````

````python
def add_mul(choice, *args): 
    if choice == "add":   # 매개변수 choice에 "add"를 입력받았을 때
        result = 0 
        for i in args: 
            result = result + i 
    elif choice == "mul":   # 매개변수 choice에 "mul"을 입력받았을 때
        result = 1 
        for i in args: 
            result = result * i 
    return result 
````

### 3) 키워드 매개변수

````python
def print_kwargs(**kwargs): 
    print(kwargs)
````

````python
print_kwargs(a=1)

print_kwargs(name='foo', age=3)
````

⚠️일반 매개변수, 가변 매개변수(```*args```), 키워드 매개변수(```**kwargs```)를 함께 사용할 때에는 반드시 순서를 맞춰야 한다.

````python
def mixed_function(name, *args, **kwargs):
    print(f"이름: {name}")
    print(f"추가 인수들: {args}")
    print(f"키워드 인수들: {kwargs}")
````
