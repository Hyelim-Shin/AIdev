# Part 1. Colab 접속하기

## 1-1. Colab이란?

정식 명칭은 **Colaboratory**. 구글이 제공하는 **클라우드 기반의 무료 파이썬 노트북 실행 환경**입니다.
내 컴퓨터에 파이썬을 설치하지 않아도, **웹 브라우저와 구글 계정만으로** 바로 시작할 수 있습니다.

| 장점 | 설명 |
|:--|:--|
| **무료** | 구글 계정만 있으면 별도 비용 없이 사용 |
| **클라우드 기반** | 파이썬·패키지 설치 등 로컬 환경 구축이 불필요 |
| **GPU / TPU 지원** | 딥러닝처럼 연산이 많은 작업을 가속 (요금제에 따라 사용량 차등) |
| **협업** | 구글 문서처럼 링크 공유 → 실시간 공동 작업 |
| **Drive 연동** | 내 Google Drive의 파일을 바로 읽고 쓰기 |

> 정리하면 **"설치 없이 바로 쓰는 노트북 + 무료 GPU"** 가 Colab을 쓰는 이유입니다.

# Part 2. 노트북(.ipynb) 다루기

## 2-1. 지금 보고 있는 이 파일의 정체

우리가 지금까지 본 파이썬 파일은 `.py` 였습니다. 파일 전체를 한 번에 실행하고, 결과는 터미널에 텍스트로 나옵니다.

**노트북(Notebook)** 은 코드를 **여러 조각(셀)으로 나누어 한 조각씩 실행**하고, 그 결과를 **코드 바로 아래에** 붙여 두는 문서입니다.
정식 명칭은 **Jupyter Notebook**이며, 파일 확장자는 `.ipynb` (**I**nteractive **Py**thon **N**ote**b**ook)입니다.

| | `.py` 스크립트 | `.ipynb` 노트북 |
|:--|:--|:--|
| 실행 단위 | 파일 전체 | **셀 하나씩** |
| 결과 확인 | 터미널 출력 | 코드 **바로 아래**에 표시 |
| 그래프·표 | 별도 창 / 파일 저장 | 문서 안에 **그대로** 삽입 |
| 설명 작성 | 주석(`#`)만 | **Markdown 텍스트 셀** |
| 적합한 일 | 완성된 프로그램, 자동화 | **탐색·실험·학습·발표** |

데이터 분석과 머신러닝의 **초반 탐색 단계**는 "한 번에 완성"되지 않습니다.
데이터를 조금 보고 → 그래프를 그려 보고 → 모델을 바꿔 보고 → 다시 확인하는 **반복 실험**이라서,
중간 결과를 바로 눈으로 확인할 수 있는 노트북이 편합니다. 이 수업의 실습을 노트북으로 하는 이유도 그것입니다.

### ⚠️ 노트북이 만능은 아닙니다

노트북은 **탐색·학습·발표에 맞는 도구**일 뿐, 모든 상황에 쓰는 것이 아닙니다. 약점이 분명합니다.

* 셀을 아무 순서로나 실행할 수 있어서 **재현이 어렵습니다** (2-4에서 직접 확인합니다)
* 파일 내부가 JSON이라 **버전 관리(diff)와 코드 리뷰가 불편**합니다
* 테스트, 모듈 재사용, 자동 실행(스케줄링)에는 맞지 않습니다

그래서 실제로는 **탐색은 노트북으로 하되, 자리를 잡은 코드는 `.py` 모듈로 옮겨** 쓰는 경우가 많습니다.
학습 파이프라인이나 서비스 코드를 노트북 상태 그대로 운영하는 일은 드뭅니다.

## 2-2. 코드 셀 실행하기

노트북은 **셀(cell)의 모음**입니다. 종류는 두 가지뿐입니다.

* **코드 셀 (Code cell)** : 파이썬 코드를 작성하고 실행하는 칸
* **텍스트 셀 (Text cell)** : Markdown 문법으로 설명을 쓰는 칸 (지금 읽고 있는 이 칸)

코드 셀을 실행하는 방법은 두 가지입니다.

* 셀 왼쪽의 **▶ 실행 버튼** 클릭
* **`Shift + Enter`** (실행 후 다음 셀로 이동)

▶️ 직접 해 보세요.

`````python
print("Hello, Colab!")
`````

노트북의 편리한 점 하나 더 — **셀의 마지막 줄에 있는 값은 `print` 없이도 출력**되며,
표나 그래프 같은 결과물은 **그림 그대로** 표시됩니다.

`````python
2 ** 10   # print 없이도 마지막 줄의 값은 화면에 나옵니다
`````

`````python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 2 * np.pi, 100)
plt.plot(t, np.sin(t))
plt.title("sin(t)")
plt.show()   # 그래프가 이 셀 바로 아래에 그려집니다
`````

## 2-3. 커널(kernel) — 셀들은 기억을 공유합니다

**커널**은 셀에 적힌 코드를 실제로 실행해 주는 **파이썬 프로세스**입니다. (Colab에서는 런타임이 곧 커널입니다.)

핵심은 **노트북의 모든 셀이 하나의 커널을 공유한다**는 점입니다.
그래서 위 셀에서 만든 변수를 아래 셀에서 그대로 쓸 수 있고,
반대로 **커널을 다시 시작하면 모든 변수가 한꺼번에 사라집니다.**

▶️ 아래 두 셀을 순서대로 실행해 보세요.

`````python
message = "Hello, Colab!"
print(message)
`````

`````python
# 위 셀에서 만든 변수를 그대로 사용할 수 있습니다.
print(message.upper())
print("글자 수 :", len(message))
`````

## 2-4. ⚠️ 셀은 "쓴 순서"가 아니라 "실행한 순서"대로 기억합니다

셀 왼쪽의 `[1]`, `[2]` 숫자는 **실행 순서**입니다. 위에서 아래로 배치한 순서가 아닙니다.
중간 셀만 여러 번 고쳐 실행하면 변수 상태가 꼬여서, **코드는 맞는데 결과가 이상한** 상황이 생깁니다.

▶️ 아래 두 셀을 실행하고, **두 번째 셀만 여러 번 반복 실행**해 보세요.

`````python
x = 10
print("x =", x)
`````

`````python
x = x + 5   # 이 셀을 여러 번 실행해 보세요. 실행할 때마다 값이 계속 커집니다!
print("x =", x)
`````

결과가 이상하면 **처음부터 깨끗하게** 돌려보는 것이 정답입니다.

## 2-5. 텍스트 셀과 Markdown

텍스트 셀은 **더블 클릭**하면 편집 모드가 되고, 다 쓰고 나서 `Shift + Enter`를 누르면 서식이 적용된 모습으로 바뀝니다.
(Colab은 편집 중 오른쪽에 미리보기를 함께 보여 줍니다.)

### Markdown 기본 문법

| 요소 | 문법 | 결과 |
|:--|:--|:--|
| 제목 | `# 제목1` `## 제목2` `### 제목3` | 크기가 다른 제목 |
| 강조 | `*기울임*` `**굵게**` `***굵은 기울임***` | *기울임* **굵게** ***굵은 기울임*** |
| 목록 | `* 항목` / `1. 항목` | 글머리표 · 번호 목록 |
| 링크 | `[표시할 글](URL)` | [부산대학교](https://www.pusan.ac.kr) |
| 이미지 | `![설명](이미지 URL)` | 이미지 삽입 |
| 인용 | `> 인용 내용` | 들여쓴 인용문 |
| 코드(한 줄) | `` `code` `` | `code` |
| 코드 블록 | ```` ```python … ``` ```` | 문법 강조된 코드 상자 |
| 표 | `\| A \| B \|` + `\|:--\|:--\|` | 지금 보고 있는 이 표 |
| 수식 | `$y = wx + b$` | $y = wx + b$ |

> 💡 **연습** : 이 셀을 더블 클릭해서 위 표가 실제로 어떻게 쓰여 있는지 직접 확인해 보세요.

## 3-3. Google Drive 마운트

### 마운트(mount)란?
저장 장치를 운영체제의 파일 시스템에 **연결해서 쓸 수 있게 만드는 것**.
(예: USB를 꽂으면 `E:` 드라이브로 보이는 것)

### 왜 필요한가?
Colab의 저장 공간(`/content`)은 **일회용**입니다. 런타임이 끊기면 방금 만든 `sample.txt` 도 **사라집니다.**
데이터셋과 학습 결과를 **내 Google Drive에 저장**해 두어야 다음에 다시 쓸 수 있습니다.

🖱️ **방법 1 (버튼)** : 왼쪽 사이드바 📁 파일 아이콘 → 드라이브 마운트 버튼 클릭

▶️ **방법 2 (코드)** : 아래 셀 실행 → 팝업에서 계정 선택 → 권한 허용

`````python
from google.colab import drive

drive.mount('/content/drive')   # 성공하면: Mounted at /content/drive

!ls /content/drive/MyDrive | head -5   # 내 드라이브 목록 중 5개만
`````

`````python
# 실습용 폴더를 만들고 그 안으로 이동하기
!mkdir -p /content/drive/MyDrive/AIdev_practice   # -p : 이미 있어도 에러 안 냄

# ⚠️ !cd 가 아니라 %cd 입니다.
#    ! 로 실행한 명령은 그 줄에서만 유효해서 폴더 이동이 유지되지 않습니다.
%cd /content/drive/MyDrive/AIdev_practice

!pwd   # 현재 위치 확인 (print working directory)
`````

✅ **체크포인트 3** — 웹브라우저에서 내 Google Drive를 열었을 때 `AIdev_practice` 폴더가 보이나요?

> 💡 `!` 와 `%` 의 차이
> * `!명령` : 리눅스 셸 명령을 **한 줄짜리 새 프로세스**로 실행 (`!ls`, `!pip install`, `!nvidia-smi`)
> * `%명령` : 노트북 자신에게 내리는 **매직 명령** (`%cd`, `%time`, `%matplotlib inline`)
> 폴더 이동처럼 **상태가 남아야 하는 일**은 반드시 `%cd` 를 써야 합니다.

| Colab | Jupyter | 기능 |

| `Ctrl + M, B` | `Esc` → `B` | 현재 셀 **아래**에 새 셀 추가 (Below) |
| `Ctrl + M, D` | `Esc` → `D, D` | 현재 셀 **삭제** (Delete) |
| `Ctrl + M, M` | `Esc` → `M` | 코드 셀 → **텍스트 셀**로 변경 (Markdown) |
| `Ctrl + M, Y` | `Esc` → `Y` | 텍스트 셀 → **코드 셀**로 변경 |
| `Ctrl + M, Z` | `Esc` → `Z` | 실행 취소 (셀 삭제 되돌리기) |
