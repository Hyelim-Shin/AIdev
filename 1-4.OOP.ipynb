## Class
**객체 지향 프로그래밍 (Object-Oriented Programming, OOP)**
객체 지향 프로그래밍은 데이터와 그 데이터를 조작하는 함수(메서드)를 하나의 객체로 묶어서 관리하는 프로그래밍 패러다임이다.

Class는 객체 지향 프로그래밍을 위한 python의 문법적 도구

자동차를 class로 구현한다면?
```python
class 자동차:
    # 자동차의 속성
    색상
    속도
    # 자동차의 기능
    속도올리기()
    속도내리기()
```

````python
class Car:
    # 자동차의 필드
    color = ""
    speed = 0

    # 자동차의 메서드(함수)
    def upSpeed(self, inc):
        self.speed += inc
    
    def downSpeed(self, dec):
        self.speed -= dec
````

````python
my_car = Car()
your_car = Car()

my_car.color = "white"
your_car.color = "black"
# my_car.speed = 0
# your_car.speed = 0

my_car.upSpeed(30)
your_car.upSpeed(60)

print(f"Color of my car is {my_car.color} and the present speed of my car is {my_car.speed}")
print(f"Color of your car is {your_car.color} and the present speed of your car is {your_car.speed}")
````

### 생성자
인스턴스를 생성하면서 필드값을 초기화시키는 함수

````python
class Car:
    # 생성자
    def __init__(self):
        self.color = "white"
        self.speed = 0
    # 자동차의 메서드(함수)
    def upSpeed(self, inc):
        self.speed += inc
    
    def downSpeed(self, dec):
        self.speed -= dec
````

````python
class Car:
    # 생성자 with argument
    def __init__(self, color):
        self.color = color
        self.speed = 0
    # 자동차의 메서드(함수)
    def upSpeed(self, inc):
        self.speed += inc
    
    def downSpeed(self, dec):
        self.speed -= dec
````

````python
my_car = Car("white")
your_car = Car("black")
````

#### 클래스 변수 vs 인스턴스 변수

````python
class Dog:
    tricks = []          # 클래스 변수 (공유됨!)
    def add_trick(self, t): self.tricks.append(t)
a, b = Dog(), Dog()
a.add_trick("앉아")
print(b.tricks)
````

````python
class Dog:
    def __init__(self):
        self.tricks = []  # 인스턴스 변수 (공유되지 않음!)
    def add_trick(self, t):
        self.tricks.append(t)
a, b = Dog(), Dog()
a.add_trick("앉아")
print(b.tricks)
````

## 상속

````python
class Person:
    # 눈 두 개, 코 하나, 입 하나...
    eyes = 2
    nose = 1
    mouth = 1
    ears = 2
    arms = 2
    legs = 2

    # 먹고 자고 이야기하고...
    def eat(self):
        print('얌냠...')

    def sleep(self):
        print('쿨쿨...')

    def talk(self):
        print('주절주절...')
````

````python
class Student(Person):     # Person 클래스를 상속받음
    def study(self):    # 메서드 오버라이딩
        print('열공열공...')
````

### 1단계. `Account` 클래스를 만드세요.
 
- 생성자는 `owner`(예금주 이름)와 `balance`(초기 잔액)를 받습니다. 초기 잔액을 안 넘기면 `0`으로 시작합니다.
- 필드: `owner`, `balance`, `history`(거래 기록 리스트, 처음엔 비어 있음)
- 메서드:
  - `deposit(amount)`: 잔액에 더하고, `history`에 `"입금 +amount"` 형태로 기록
  - `withdraw(amount)`: 잔액이 부족하면 `"잔액이 부족합니다"`를 출력하고 아무것도 하지 않음. 충분하면 빼고 `"출금 -amount"` 기록
  - `show()`: 예금주와 현재 잔액 출력
  - `show_history()`: 거래 기록을 한 줄씩 번호를 붙여 출력
**실행 예시**
 
```python
acc = Account("김철수", 10000)
acc.deposit(5000)
acc.withdraw(3000)
acc.withdraw(50000)
acc.show()
acc.show_history()
```
