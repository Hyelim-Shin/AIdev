**숫자형을 위한 연산자**

````python
## 더하기, 빼기, 곱하기, 나누기
a = 3; b = 4; c = 6
print(f'a = {a}, b = {b}, c = {c}')
print(f'a+b = {a+b}')
print(f'b-c = {b-c}')
print(f'a*b = {a*b}')
print(f'b/a = {b/a}')
print(f'c/a = {c/a}')
````

````python
## 몫, 나머지, 제곱
print(f'b/a 몫: {b//a}')
print(f'b/a 나머지: {b%a}')
print(f'b^a = {b**a}')
````

**복합 연산자**

산술 연산자(```+```,```-``` 등)와 대입 연산자(```=```)를 합친 것으로 복합 연산자에는 다음과 같은 것들이 있다.

```python
+=, -=, *=, /=, //=, %=, **=
```

````python
a = 1
a = a + 1
print(a)
````

````python
a += 1
print(a)
````

````python
a -=1
print(a)
````
