### 사용 방법

#### 첫번째 방법

```python
import 모듈_이름

모듈_이름.모듈_함수()
a = 모듈_이름.모듈_함수()
```

````python
import calendar

calendar.prmonth(2025, 9)
````

#### 두번째 방법

```python
from 모듈_이름 import 모듈_함수

모듈_함수()
a = 모듈_함수()
```


````python
del calendar
calendar.prmonth(2025, 9)
````

#### 두번째 방법 확장1

```python
from 모듈_이름 import 모듈_함수1, 모듈_함수2
```

````python
from calendar import prmonth, isleap
prmonth(2024, 2)
y2024_is_leap = isleap(2024)

prmonth(2025, 2)
y2025_is_leap = isleap(2025)

print(f'\n\n2024 is leap year? {y2024_is_leap}')
print(f'2025 is leap year? {y2025_is_leap}')
````

#### 두번째 방법 확장2

```python
from 모듈_이름 import *
```

````python
del prmonth
del isleap
prmonth(2024, 2)
````

````python
from calendar import *
prmonth(2024, 2)
y2024_is_leap = isleap(2024)

prmonth(2025, 2)
y2025_is_leap = isleap(2025)

print(f'\n\n2024 is leap year? {y2024_is_leap}')
print(f'2025 is leap year? {y2025_is_leap}')
````

#### ⚠️ `import *` 는 권장하지 않음

* 모듈 안에 있는 **모든** 이름이 한 번에 들어오기 때문에
    * 어떤 함수가 어느 모듈에서 왔는지 코드만 보고 알 수 없음
    * 내가 만든 함수(변수)와 이름이 같으면 **조용히 덮어써짐**
* 아래 셀: 내가 만든 `isleap` 함수가 `from calendar import *` 이후 calendar의 `isleap`으로 바뀜

````python
def isleap(year):
    return f'{year}년: 내가 만든 isleap 함수'

print(isleap(2024))

from calendar import *   # calendar 안의 isleap이 내 isleap을 덮어씀

print(isleap(2024))
````

#### 세번째 방법: 별칭(alias) 붙이기

```python
import 모듈_이름 as 별칭

별칭.모듈_함수()
```

* 모듈 이름이 길거나 자주 쓸 때 짧은 이름으로 줄여서 사용
* `from 모듈_이름 import 모듈_함수 as 별칭` 도 가능

````python
import calendar as cal

cal.prmonth(2025, 9)
print(cal.isleap(2024))
````

````python
from calendar import isleap as is_leap_year

print(is_leap_year(2024))
````

#### 자주 보게 될 별칭 (관례)

| 코드 | 별칭 |
|---|---|
| `import numpy as np` | `np` |
| `import pandas as pd` | `pd` |
| `import matplotlib.pyplot as plt` | `plt` |
| `import torch.nn as nn` | `nn` |

* 문법적으로 강제는 아니지만, 거의 모든 코드가 이 별칭을 쓰기 때문에 **그대로 따르는 것이 좋음**
* 다른 사람 코드를 읽을 때 `np.`, `pd.`, `plt.` 가 보이면 어떤 라이브러리인지 바로 알 수 있음

### 직접 모듈 만들어 보기

* 셀 맨 위에 `%%writefile 파일이름.py` 를 쓰면 그 셀의 내용이 **파일로 저장**됨 (코랩 전용 명령, 파이썬 문법 아님)
* 실행 후 왼쪽 📁 파일 탭에서 `mod1.py` 가 생긴 것을 확인해 보기
* 코랩 런타임이 끊기면 파일도 사라지므로, 노트북을 다시 열었다면 이 셀부터 다시 실행

````python
%%writefile mod1.py
def add(a, b):
    return a + b

def sub(a, b):
    return a - b
````

* 이제 `mod1.py` 는 하나의 모듈 → `import mod1` 으로 불러와서 사용 가능

````python
import mod1

print(mod1.add(3, 4))
print(mod1.sub(3, 4))
````

* 파일 내용을 고쳤다면?
    * 파이썬은 한 번 불러온 모듈을 기억(캐시)하기 때문에 `import mod1` 을 다시 해도 바뀐 내용이 반영되지 않음
    * `importlib.reload(mod1)` 을 쓰거나, 런타임을 재시작해야 함

````python
%%writefile mod1.py
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):   # 새로 추가한 함수
    return a * b
````

````python
import mod1
# mod1.mul(3, 4)  # 주석을 풀고 실행하면 AttributeError: 아직 예전 mod1을 기억하고 있음

import importlib
mod1 = importlib.reload(mod1)
print(mod1.mul(3, 4))
````
